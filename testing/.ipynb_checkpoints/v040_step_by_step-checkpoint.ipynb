# v0.4.0 step-by-step: how the predicted light builds up in one TPC

Runs the **full v0.4.0 chain on the whole event** (algorithm untouched) and pauses
**every time the watched TPC's predicted light (base image) changes**, showing:
1. the usual 2D light+charge display (`plot_tpc_light_and_charge`, actual vs predicted), and
2. an interactive 3D reconstructed display (hits colored by current t0 group; **gray = no t0 yet**).

Stage banners say what is happening; nothing is saved to disk. Timings for every stage are
collected and summarized at the end (including stages that never touch this TPC).

**Kernel**: use the `module load pytorch/2.8.0` python (the ~/.local torch is broken for the renderer).
Default subject: `0000003 ev6 TPC 12` — 650 MeV, 23 clusters, 98% efficient (moderate, imperfect).

```python
import sys, time
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

TESTING = "/global/cfs/cdirs/dune/users/yuxuan/NDLAr-full/CLMatching_v0.1_backboneOnly/testing"
sys.path.insert(0, TESTING)
import v040_walkthrough as W

# ---- choose the subject (moderate energy, reasonably okay) ----
FTAG, EV, TPC = "0000003", 6, 12
DATA = ("/global/cfs/cdirs/dunepro/people/abooth/nd-production/output/MiniProdN5/"
        "run-ndlar-flow/MiniProdN5p1_NDComplex_FHC.flow.full.sanddrift/FLOW/0000000")
FP = f"{DATA}/MiniProdN5p1_NDComplex_FHC.flow.full.sanddrift.{FTAG}.FLOW.hdf5"

from ML_NDfull_perceiver import load_tpc_geometries, DEFAULT_TPC_YAML
from visualizationTools import plot_tpc_light_and_charge
import plottingTools as pt
geom_map = load_tpc_geometries(DEFAULT_TPC_YAML)
print(f"subject: {FTAG} ev{EV} TPC {TPC}")
```

```python
# heavy setup: models + v0.4.0 config (boundary 30, predict >=1.5 MeV, veto-4, vertex merge)
ctx = W.load_chain()
```

## Stage 0 — first stage (clustering + vertex merge + prediction + shower/track placement)
This runs the production first stage for the WHOLE event. Afterwards we print the
**cluster definition of the watched TPC** and show the initial state: the backbone base image
already contains phase-1 placements (we rebuild it step-by-step in Stage 1 below, so the
walkthrough displays start from an empty TPC).

```python
S = W.first_stage(ctx, FP, EV, TPC)

# ---- cluster definition for this TPC ----
import pandas as pd
rows, n_noise = W.cluster_table(S)
df = pd.DataFrame(rows).sort_values("e_tpc", ascending=False)
print(f"clusters with hits in TPC {TPC}: {len(df)} (+ {n_noise} noise hits with no cluster)")
display(df)
```

```python
# ---- display machinery: fires on every base-image change of the watched TPC ----
_prev = {"base": None, "n": 0}

def show_state(stage, note, base_view=None, force=False):
    base = np.asarray(base_view if base_view is not None else W.current_base(S)[TPC], np.float32)
    changed = _prev["base"] is None or not np.allclose(base, _prev["base"], atol=1.0)
    if not (changed or force):
        return
    _prev["n"] += 1
    print(f"\n{'='*100}")
    print(f"  frame {_prev['n']:>3} | [{stage}]  {note}")
    print(f"{'='*100}")
    if changed:
        _prev["base"] = base.copy()
    # 1) the usual 2D light + charge display
    plot_tpc_light_and_charge(ev_id=EV, TPCid=TPC, hits_full_ev=S["hits_full_ev"],
                              actual=S["actual"][TPC].astype(np.float32),
                              predicted=base, geom_map=geom_map, individualScaling=False)
    plt.show()
    # 2) interactive 3D reconstructed display (gray = not yet assigned)
    ts = W.current_ts(S)
    m = S["hit_tpc"] == TPC
    safe = np.where(np.isfinite(ts) & (ts >= 0), ts, -1.0)
    g = pt.group_hits_by_time(safe[m], time_window=10)
    pt.plot_3d_clusters_with_t0(
        x=np.asarray(S["fs"].event.x)[m], y=np.asarray(S["fs"].event.y)[m],
        z=np.asarray(S["fs"].event.z)[m], labels=g, t0s=safe[m],
        energies=S["e_pos"][m], title=f"[{stage}] {note[:70]}", save_path=None)

# initial state: nothing placed in the walkthrough's rebuilt base yet
np.asarray(S["fs"].track_stage.base_image)[:] = 0.0
S["hit_t0"][:] = np.nan
show_state("0 initial", "empty base — no cluster has a t0 yet (all gray)", force=True)
```

## Stage 1 — stack2 backbone
Showers first, then >=3-TPC tracks, then the 1-2-TPC pool via **energy-banded permutation
batches** (cosine-softened scans, defer-once overflow gate), then the swap and rescue clean-up
passes. **A frame appears for every placement that touches this TPC.**

```python
W.stack2_backbone(S, show_state)
```

## Stage 2 — overflow trigger + saving algorithm
The locked trigger flags suspect placements; co-located groups get Stage-A anchored
permutations, all-stay groups get Stage-B part detachment (NEW cluster ids), lone flags get
the N=1 rescan/detach. **Frames appear for every ADOPTED move/detach touching this TPC.**

```python
W.rescue_stage(S, show_state)
```

## Stages 3+4 — Phase 2 (large clusters) + police, then V2 + police
Phase 2 places multi-TPC and >30 MeV clusters (its internal order is not hookable, so one
frame shows everything it added to this TPC, listed); the police probes every placement and
fixes alarms immediately. Then V2 (spatial+light rescue) with the same policing.

```python
W.staged_stage(S, show_state)
```

## Stages 5-8 — family expansion, error-matrix tail, swap fix, force coverage
Expansion frames are replayed **one absorbed cluster at a time** (family id, score, geodesic
distance annotated). The matrix, swap-fix and force stages each get a summary frame.

```python
W.smallcluster_stage(S, show_state)
```

## Final summary — timings for every stage (whole-event work) + this TPC's outcome

```python
import pandas as pd
tdf = pd.DataFrame(S["timings"], columns=["stage", "seconds"])
tdf["seconds"] = tdf["seconds"].round(1)
display(tdf)
print(f"TOTAL {tdf['seconds'].sum():.1f}s for the whole event (all TPCs)")

summ = W.tpc_summary(S)
print(f"\nwatched TPC {TPC}: truth-valid E {summ['e_truth']:.0f} MeV | "
      f"coverage {summ['coverage']*100:.1f}% | efficiency {summ['eff']*100:.2f}%")
show_state("final", "final assignment state", force=True)
```
